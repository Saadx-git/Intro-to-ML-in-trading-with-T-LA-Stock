```python
# Boilerplate
import pandas as pd
from datasets import load_dataset

data = load_dataset('codesignal/tsla-historic-prices')
tesla_df = pd.DataFrame(data['train'])

tesla_df['Date'] = pd.to_datetime(tesla_df['Date'])
tesla_df.set_index('Date', inplace=True)
tesla_df.sort_index(inplace=True)


# TODO: Create 'Volatility_Percentage'
# Formula: (High - Low) / Open
tesla_df['Volatility_Percentage'] = (tesla_df['High'] - tesla_df['Low']) / tesla_df['Open']


# TODO: Create 'Momentum_Percentage'
# Formula: (Close - Open) / Open
tesla_df['Momentum_Percentage'] = (tesla_df['Close'] - tesla_df['Open']) / tesla_df['Open']


# TODO: Create 'Strong_Bull_Day'
# Condition:
# 1 if Momentum_Percentage > 0.02 else 0
tesla_df['Strong_Bull_Day'] = (tesla_df['Momentum_Percentage'] > 0.02).astype(int)


# TODO: Print first 5 rows of:
# ['Volatility_Percentage', 'Momentum_Percentage', 'Strong_Bull_Day']
print(tesla_df[['Volatility_Percentage', 'Momentum_Percentage', 'Strong_Bull_Day']].head(5))
```
