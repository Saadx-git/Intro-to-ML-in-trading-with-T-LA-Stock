```python
# Boilerplate: Import libraries and load Tesla dataset
import pandas as pd
import matplotlib.pyplot as plt
from datasets import load_dataset

dataset = load_dataset('codesignal/tsla-historic-prices')
tesla_df = pd.DataFrame(dataset['train'])

tesla_df['Date'] = pd.to_datetime(tesla_df['Date'])
tesla_df.set_index('Date', inplace=True)
tesla_df.sort_index(inplace=True)


# TODO: Calculate 20-day EMA for 'Volume'
tesla_df['EMA_20'] = tesla_df['Volume'].ewm(span=20, adjust = False).mean()

# TODO: Calculate 50-day EMA for 'Volume'
tesla_df['EMA_50'] = tesla_df['Volume'].ewm(span=50, adjust = False).mean()

# TODO: Create 'Volume_Signal' column initialized to 0
tesla_df['Volume_Signal'] = 0

# TODO: Set Volume_Signal = 1 where EMA_20 > EMA_50
tesla_df.loc[tesla_df['EMA_20'] > tesla_df['EMA_50'] , 'Volume_Signal'] = 1

# TODO: Set Volume_Signal = -1 where EMA_20 < EMA_50

tesla_df.loc[tesla_df['EMA_20'] < tesla_df['EMA_50'] , 'Volume_Signal'] = -1

# TODO: Create 'Volume_Crossover' using diff()
tesla_df['Volume_Crossover'] = tesla_df['Volume_Signal'].diff()

# TODO: Filter dataset for year 2023
tesla_2023 = tesla_df.loc['2023']

# TODO: Plot 'Volume', 'EMA_20', 'EMA_50' using DataFrame .plot() (figsize=12,6, title='TSLA Volume EMA (2023)')
tesla_2023[['Volume', 'EMA_20', 'EMA_50']].plot(figsize=(20,10), title='TSLA Volume EMA (2023)')
plt.grid(True)


# TODO: Highlight crossover points:
#       - If Volume_Crossover == 2 → green dot
#       - If Volume_Crossover == -2 → red dot
# Loop through the 2023 filtered DataFramePreparing Financial Data for Machine Learning
for index, row in tesla_2023.iterrows():
    if row['Volume_Crossover'] == 2:
        plt.scatter(index, row['EMA_20'], color='green', marker='o', s=50, label='Bullish Crossover' if 'Bullish Crossover' not in plt.gca().get_legend_handles_labels()[1] else "")
    elif row['Volume_Crossover'] == -2:
        plt.scatter(index, row['EMA_20'], color='red', marker='o', s=50, label='Bearish Crossover' if 'Bearish Crossover' not in plt.gca().get_legend_handles_labels()[1] else "")

plt.legend()
plt.show()
```
